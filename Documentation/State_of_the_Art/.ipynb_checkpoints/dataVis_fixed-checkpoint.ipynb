# Paper Metrics Analysis
Read papers downloaded from the Springer Open API and augmented with web scraping from the SCImago website for h-index of their journals

## 1. Load and Explore Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Step 1: Read JSON
df = pd.read_json('bulkArticlesAugmented.json')

# Expand nested dictionaries into columns
normalized_df = pd.json_normalize(df['records'])

print(f"Total papers loaded: {len(normalized_df)}")
print(f"\nColumns available: {normalized_df.columns.tolist()}")
```

## 2. Basic Statistics

```python
# Check h-index data quality
print("=== H-INDEX STATISTICS ===")
print(f"Papers with h-index: {normalized_df['h_index'].notna().sum()}")
print(f"Papers without h-index: {normalized_df['h_index'].isna().sum()}")

if normalized_df['h_index'].notna().sum() > 0:
    print(f"\nH-index statistics (excluding missing values):")
    print(normalized_df['h_index'].describe())
else:
    print("\n⚠️ WARNING: No h-index values found. Run ETL.py first to fetch journal metrics.")
```

## 3. Data Preparation

```python
# Convert dates
normalized_df['onlineDate'] = pd.to_datetime(normalized_df['onlineDate'])
normalized_df['monthYear'] = normalized_df['onlineDate'].dt.to_period('M').astype(str)

# Filter valid papers (with h-index data)
df_filtered = normalized_df[normalized_df['h_index'].notna()].copy()

print(f"Papers with valid h-index data: {len(df_filtered)}")
print(f"\nAvailable columns: {list(df_filtered.columns)}")
```

## 4. Paper Details Table

```python
# Create a comprehensive summary table
# Only include columns that exist
base_cols = ['title', 'publicationName', 'h_index', 'onlineDate', 'doi', 'key_issn']
optional_cols = ['journal_name_verified', 'metrics_source']

summary_cols = [col for col in base_cols if col in df_filtered.columns]
summary_cols.extend([col for col in optional_cols if col in df_filtered.columns])

papers_summary = df_filtered[summary_cols].sort_values('h_index', ascending=False)

# Display with full text
with pd.option_context('display.max_colwidth', None, 
                       'display.max_rows', None,
                       'display.width', None):
    display(papers_summary)
```

## 5. Visualizations

```python
# H-index distribution
if len(df_filtered) > 0:
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    
    # Histogram
    axes[0].hist(df_filtered['h_index'], bins=20, color='skyblue', edgecolor='black')
    axes[0].set_title('Distribution of Journal H-Index', fontsize=14, fontweight='bold')
    axes[0].set_xlabel('H-Index', fontsize=12)
    axes[0].set_ylabel('Number of Papers', fontsize=12)
    axes[0].grid(axis='y', alpha=0.3)
    
    # Box plot
    axes[1].boxplot(df_filtered['h_index'], vert=True)
    axes[1].set_title('H-Index Box Plot', fontsize=14, fontweight='bold')
    axes[1].set_ylabel('H-Index', fontsize=12)
    axes[1].grid(axis='y', alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('h_index_distribution.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print(f"\nMean H-Index: {df_filtered['h_index'].mean():.2f}")
    print(f"Median H-Index: {df_filtered['h_index'].median():.2f}")
    print(f"Std Dev: {df_filtered['h_index'].std():.2f}")
```

```python
# Top journals by h-index
if len(df_filtered) > 0:
    top_journals = df_filtered.groupby('publicationName')['h_index'].agg(['mean', 'count']).sort_values('mean', ascending=False).head(10)
    
    fig, ax = plt.subplots(figsize=(14, 8))
    bars = ax.barh(range(len(top_journals)), top_journals['mean'], color='coral')
    ax.set_yticks(range(len(top_journals)))
    ax.set_yticklabels(top_journals.index, fontsize=10)
    ax.set_xlabel('Average H-Index', fontsize=12)
    ax.set_title('Top 10 Journals by Average H-Index', fontsize=14, fontweight='bold', pad=20)
    ax.invert_yaxis()
    
    # Add value labels
    for i, (bar, count) in enumerate(zip(bars, top_journals['count'])):
        width = bar.get_width()
        ax.text(width, bar.get_y() + bar.get_height()/2, 
                f'{width:.1f} (n={count})', 
                ha='left', va='center', fontsize=9, fontweight='bold')
    
    plt.tight_layout()
    plt.savefig('top_journals.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
# Publications over time
if len(df_filtered) > 0:
    df_sorted = df_filtered.sort_values('onlineDate')
    
    plt.figure(figsize=(12, 6))
    ax = sns.countplot(
        data=df_sorted,
        x='monthYear',
        order=df_sorted['monthYear'].unique(),
        color='steelblue'
    )
    
    # Force y-ticks to increment by 1
    max_count = df_sorted['monthYear'].value_counts().max()
    plt.yticks(np.arange(0, max_count + 1, 1))
    
    plt.title('Distribution of Selected Papers Monthly', fontsize=14, fontweight='bold', pad=20)
    plt.xlabel('Month-Year', fontsize=12)
    plt.ylabel('Number of Papers', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    
    # Add value labels on top of bars
    for p in ax.patches:
        height = p.get_height()
        if height > 0:
            ax.annotate(
                f'{int(height)}',
                (p.get_x() + p.get_width() / 2., height),
                ha='center', va='bottom',
                xytext=(0, 3),
                textcoords='offset points',
                fontsize=10,
                fontweight='bold'
            )
    
    plt.tight_layout()
    plt.savefig('publications_per_month.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
# H-index trends over time
if len(df_filtered) > 0:
    monthly_stats = df_filtered.groupby('monthYear')['h_index'].agg(['mean', 'count', 'std'])
    
    fig, ax1 = plt.subplots(figsize=(14, 6))
    
    color = 'tab:blue'
    ax1.set_xlabel('Month-Year', fontsize=12)
    ax1.set_ylabel('Average H-Index', color=color, fontsize=12)
    line1 = ax1.plot(monthly_stats.index, monthly_stats['mean'], 
                     color=color, marker='o', linewidth=2, markersize=8, label='Avg H-Index')
    ax1.tick_params(axis='y', labelcolor=color)
    ax1.grid(alpha=0.3)
    
    ax2 = ax1.twinx()
    color = 'tab:red'
    ax2.set_ylabel('Number of Papers', color=color, fontsize=12)
    line2 = ax2.plot(monthly_stats.index, monthly_stats['count'], 
                     color=color, marker='s', linewidth=2, markersize=8, 
                     linestyle='--', label='Paper Count')
    ax2.tick_params(axis='y', labelcolor=color)
    
    plt.title('Average H-Index and Paper Count Over Time', 
              fontsize=14, fontweight='bold', pad=20)
    plt.xticks(rotation=45, ha='right')
    
    # Combine legends
    lines = line1 + line2
    labels = [l.get_label() for l in lines]
    ax1.legend(lines, labels, loc='upper left')
    
    plt.tight_layout()
    plt.savefig('h_index_over_time.png', dpi=300, bbox_inches='tight')
    plt.show()
```

## 6. Export Results

```python
# Display key metrics for all papers
print("=== PAPER METRICS SUMMARY ===")
display_cols = ['title', 'publicationName', 'h_index', 'monthYear', 'key_issn', 'doi']
available_display = [col for col in display_cols if col in df_filtered.columns]

with pd.option_context('display.max_colwidth', 80):
    display(df_filtered[available_display].sort_values('h_index', ascending=False))
```

```python
# Store the selected papers for further use
if len(df_filtered) > 0:
    df_filtered.to_csv('selectedPapers.csv', index=False)
    print(f"✓ Exported {len(df_filtered)} papers to selectedPapers.csv")
    
    # Also export a summary table (only with existing columns)
    export_cols = ['title', 'publicationName', 'h_index', 'onlineDate', 'doi', 'key_issn']
    
    # Add optional columns if they exist
    optional_export = ['journal_name_verified', 'metrics_source']
    for col in optional_export:
        if col in df_filtered.columns:
            export_cols.append(col)
    
    summary_export = df_filtered[export_cols].copy()
    
    # Try to export to Excel, fallback to CSV if openpyxl not installed
    try:
        summary_export.to_excel('selectedPapers_summary.xlsx', index=False)
        print(f"✓ Exported summary to selectedPapers_summary.xlsx")
    except ImportError:
        print("⚠️  openpyxl not installed, exporting to CSV instead")
        summary_export.to_csv('selectedPapers_summary.csv', index=False)
        print(f"✓ Exported summary to selectedPapers_summary.csv")
else:
    print("⚠️ No papers with valid h-index data to export")
```

## 7. Final Statistics Report

```python
print("="*60)
print("FINAL ANALYSIS REPORT")
print("="*60)
print(f"\nTotal papers analyzed: {len(normalized_df)}")
print(f"Papers with h-index data: {len(df_filtered)}")
print(f"Papers without h-index data: {len(normalized_df) - len(df_filtered)}")

if len(df_filtered) > 0:
    print(f"\nH-Index Statistics:")
    print(f"  - Mean: {df_filtered['h_index'].mean():.2f}")
    print(f"  - Median: {df_filtered['h_index'].median():.2f}")
    print(f"  - Min: {df_filtered['h_index'].min():.0f}")
    print(f"  - Max: {df_filtered['h_index'].max():.0f}")
    print(f"  - Std Dev: {df_filtered['h_index'].std():.2f}")
    
    print(f"\nTop 5 Papers by Journal H-Index:")
    top5 = df_filtered.nlargest(5, 'h_index')[['title', 'publicationName', 'h_index']]
    for idx, row in top5.iterrows():
        print(f"  {row['h_index']:.0f} - {row['title'][:60]}...")
        print(f"      Journal: {row['publicationName']}")
    
    print(f"\nUnique Journals: {df_filtered['publicationName'].nunique()}")
    print(f"Date Range: {df_filtered['onlineDate'].min().strftime('%Y-%m-%d')} to {df_filtered['onlineDate'].max().strftime('%Y-%m-%d')}")
else:
    print("\n⚠️ No valid h-index data found.")
    print("Please run ETL.py to fetch journal metrics from SCImago.")

print("\n" + "="*60)
```
